```python

def fv_from_row(row):
    """Returns a key value pair for each item in a Series"""
    cols = list(row.axes[0])
    fvs = []
    for i, v in enumerate(row):
        fvs.append((cols[i], v))

    return fvs

# fv_from_row(new_df.iloc[0])
# pd.Series.axes
```

```python


updates = []
for i, row in new_df.iterrows():
    filter = {'_id': row['_id']}
    update = {
        '$set': {
            field: value for field, value in fv_from_row(row)
        }
    }
    updates.append(UpdateOne(filter, update))

db_operations.update_table('main_table', 'all_events',  updates)
```
